```python
from xml.etree.ElementTree import parse

def getAccessionGOTerms(xmlFile, accession):
    tree = parse(xmlFile)
    root = tree.getroot()
    ns = '{http://uniprot.org/uniprot}'
    
    match_go_terms = []
    proteins = root.findall(ns + 'entry')
    for entry in proteins:
        accessions = entry.findall(ns + "accession")
        current_accessions = [acc.text for acc in accessions]
        if accession not in current_accessions:
            continue

        goTerms = entry.findall(ns + 'dbReference[@type="GO"]')
        for goT in goTerms:
            gID = goT.attrib['id']
            gName = goT.find(ns + 'property[@type="term"]').attrib['value']
            match_go_terms.append((gID, gName))
        break
    return match_go_terms
```

```python
xml_file = "data/uniprot-proteome_UP000000625.xml"

go_dict = {}

for acc in overabundant_accessions:
    go_terms = getAccessionGOTerms(xml_file, acc)
    for go_id, go_name in go_terms:
        if go_id not in go_dict:
            go_dict[go_id] = {
                "ID": go_id,
                "name": go_name,
                "carried_by": []
            }
        go_dict[go_id]["carried_by"].append(acc)
list(go_dict.items())[:5]
```

```python
for go_id in go_dict:
    go_dict[go_id]["carried_by"] = sorted(set(go_dict[go_id]["carried_by"]))

go_summary = pd.DataFrame([
    {
        "GO_ID": v["ID"],
        "GO_name": v["name"],
        "count": len(v["carried_by"]),
        "carried_by": ", ".join(v["carried_by"])
    }
    for v in go_dict.values()
]).sort_values("count", ascending=False)

go_summary.head(20)
```

```python
pval_threshold = -np.log10(0.001)
lfc_threshold = df_clean[lfc_col].mean()

df_sign = df_clean.loc[
    (df_clean[mlog10p_col] > pval_threshold) &
    (df_clean[lfc_col] > lfc_threshold)
].copy()

overabundant_accessions = df_sign[acc_col].dropna().unique()

go_dict = {}
for acc in overabundant_accessions:
    for go_id, go_name in getAccessionGOTerms("data/uniprot-proteome_UP000000625.xml", acc):
        if go_id not in go_dict:
            go_dict[go_id] = {"ID": go_id, "name": go_name, "carried_by": []}
        go_dict[go_id]["carried_by"].append(acc)

for go_id in go_dict:
    go_dict[go_id]["carried_by"] = sorted(set(go_dict[go_id]["carried_by"]))
```
