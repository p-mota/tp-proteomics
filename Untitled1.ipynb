```python
import numpy as np
import matplotlib.pyplot as plt

lfc_col = "Log2 Corrected Abundance Ratio"
mlog10p_col = "-LOG10 Adj.P-val"

pval_threshold = -np.log10(0.001)

mu = df_clean[lfc_col].mean()
sigma = df_clean[lfc_col].std()
lfc_threshold = mu + sigma

fig, ax = plt.subplots(figsize=(7, 6))

ax.scatter(
    df_clean[lfc_col],
    df_clean[mlog10p_col],
    s=10,
    alpha=0.6,
    color="grey"
)

significant = (
    (df_clean[lfc_col] > lfc_threshold) &
    (df_clean[mlog10p_col] > pval_threshold)
)

ax.scatter(
    df_clean.loc[significant, lfc_col],
    df_clean.loc[significant, mlog10p_col],
    s=25,
    alpha=0.9,
    color="red"
)

ax.axvline(lfc_threshold, color="blue", linestyle="--")
ax.axhline(pval_threshold, color="blue", linestyle="--")

ax.set_xlabel("Log2 Corrected Abundance Ratio")
ax.set_ylabel("-log10 adjusted p-value")
ax.set_title("Volcano plot")

fig

fig.savefig("figures/volcano_plot.png", dpi=300, bbox_inches="tight")
```

```python
from xml.etree.ElementTree import parse, dump

def getAccessionGOTerms(xmlFile, accession):
    tree = parse(xmlFile)
    root = tree.getroot()
    ns = '{http://uniprot.org/uniprot}'
    
    match_go_terms = []
    proteins = root.findall(ns + 'entry')
    for entry in proteins:
        accessions = entry.findall(ns+"accession")
        current_accessions = [ acc.text for acc in accessions ]
        if not accession in current_accessions:
            continue
        goTerms = entry.findall('{http://uniprot.org/uniprot}dbReference[@type="GO"]')
        #goTerms = xmlEntry.findall(ns +'dbReference[@type="GO"]')
        for goT in goTerms:
            gID   = goT.attrib['id']
            gName = goT.find(ns +'property[@type="term"]').attrib['value']
            match_go_terms.append((gID, gName))
        break
    return match_go_terms
getAccessionGOTerms("./data/uniprot-proteome_UP000000625.xml", "P0A8V6")
```

```python
xml_file = "data/uniprot-proteome_UP000000625.xml"

go_dict = {}

for acc in overabundant_accessions:
    go_terms = getAccessionGOTerms(xml_file, acc)
    for go_id, go_name in go_terms:
        if go_id not in go_dict:
            go_dict[go_id] = {
                "ID": go_id,
                "name": go_name,
                "carried_by": []
            }
        go_dict[go_id]["carried_by"].append(acc)
list(go_dict.items())[:5]
```

```python
for go_id in go_dict:
    go_dict[go_id]["carried_by"] = sorted(set(go_dict[go_id]["carried_by"]))

go_summary = pd.DataFrame([
    {
        "GO_ID": v["ID"],
        "GO_name": v["name"],
        "count": len(v["carried_by"]),
        "carried_by": ", ".join(v["carried_by"])
    }
    for v in go_dict.values()
]).sort_values("count", ascending=False)

go_summary.head(20)
```

```python
pval_threshold = -np.log10(0.001)
lfc_threshold = df_clean[lfc_col].mean()

df_sign = df_clean.loc[
    (df_clean[mlog10p_col] > pval_threshold) &
    (df_clean[lfc_col] > lfc_threshold)
].copy()

overabundant_accessions = df_sign[acc_col].dropna().unique()

go_dict = {}
for acc in overabundant_accessions:
    for go_id, go_name in getAccessionGOTerms("data/uniprot-proteome_UP000000625.xml", acc):
        if go_id not in go_dict:
            go_dict[go_id] = {"ID": go_id, "name": go_name, "carried_by": []}
        go_dict[go_id]["carried_by"].append(acc)

for go_id in go_dict:
    go_dict[go_id]["carried_by"] = sorted(set(go_dict[go_id]["carried_by"]))
```

```python
go_terms_bg = go_bg["go_terms"]

N = sum(v["count"] for v in go_terms_bg.values())

from scipy.stats import hypergeom
import pandas as pd

go_terms_bg = go_bg["go_terms"]

N = sum(v["count"] for v in go_terms_bg.values())
n = len(overabundant_accessions)

results = []

for go_id, go_info in go_dict.items():
    k = len(go_info["carried_by"])
    info_bg = go_terms_bg.get(go_id)
    if info_bg is None or k == 0:
        continue

    K = info_bg["count"]

    rv = hypergeom(N, K, n)
    pval = rv.sf(k - 1)

    results.append({
        "GO_ID": go_id,
        "GO_name": go_info["name"],
        "occurence": k,
        "pvalue": pval
    })

go_pvalues = pd.DataFrame(results).sort_values("pvalue")
go_pvalues.head(20)
```
